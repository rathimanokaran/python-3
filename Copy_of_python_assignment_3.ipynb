```python
##implement the guessing logic##
import random

secret_number = random.randint(1, 10)

attempts = 3

while attempts > 0:
    guess = int(input("Enter the correct number: "))

    if guess == secret_number:
        print("correct your win")
        break

    else:
        attempts -= 1
```
